## 🎈 Ejercicio: Arkansas 🎈


Queremos obtener de la página de Wikipedia de "Arkansas", https://es.wikipedia.org/wiki/Arkansas, la tabla que nos indica la composición étnica a aprtir del censo de EEUU de 2020.

- Revisar la página web de Wikipedia y obtener la tabla en un df

```python
url = 'https://es.wikipedia.org/wiki/Arkansas'
```

```python
arkansas_raw.info()
```

```python
#Eliminamos las columnas Alone.1 y Total.1 porque no me aportan nada.
arkansas_filt=arkansas_raw.drop(['Alone.1','Total.1'], axis=1)
arkansas_filt
```

```python
columnas=['Raza y etnia', 'Solo', 'Total']
arkansas_filt.columns=columnas
arkansas_filt
```

```python
arkansas_filt = arkansas_filt.replace({'%': ''}, regex=True)
arkansas_filt
arkansas_filt.info()
```

```python
columnas=list(arkansas_filt.columns)
columnas_filtradas=columnas[1:]
columnas_filtradas

arkansas_filt[columnas_filtradas] =arkansas_filt[columnas_filtradas].apply(pd.to_numeric, errors='coerce')
arkansas_filt.info()
arkansas_filt
```

## 🎈 Ejercicio pandas 🎈


- Pandas: scrapear la tabla que contiene informacion sobre los álbums más vendidos por año en todo el mundo (https://es.wikipedia.org/wiki/Anexo:%C3%81lbumes_musicales_m%C3%A1s_vendidos) generando un dataframe con la información del año, nombre del álbum, artista y ventas. Ésta ultima columna deberá tener sus valores en numeros enteros en millones.

```python
url='https://es.wikipedia.org/wiki/Anexo:%C3%81lbumes_musicales_m%C3%A1s_vendidos'
albums=pd.read_html(url,match='2002')
df=albums[1]
df.head()
```

```python
df_limpio=df.drop('Ref(s)',axis=1)
columnas=['Año','Album', 'Artista', 'Ventas']
df_limpio.columns=columnas
df_limpio.head()
```

```python
def millon(valor):
  return int(valor*1000000)

df_limpio['Ventas'] = df_limpio['Ventas'].apply(millon)
df_limpio.head(10)
```

## 🎈 Ejercicio beautiful soup🎈


- Beautiful Soup: scrapear la tabla de lenguajes de programación más utilizados (https://www.tiobe.com/tiobe-index/) creando un dataframe con una columna llamada "lenguaje" y otra columna llamada "porcentaje_uso" (cogiendo la info de programming language y ratings respectivamente).

```python
#Hacerlo creando un diccionario, en el ratings intentar no coger el %.
# URL de la págin
url = 'https://www.tiobe.com/tiobe-index/'

# Realizar la solicitud GET a la página
response = requests.get(url)
soup = bs(response.text, 'html.parser')
soup
tables = soup.find_all('table', {'class': 'table table-striped table-top20'})
table=tables[0]
# rows almacena todas las filas (<tr>: table rows) de la tabla encontrada (target_table)
rows = table.find_all('tr')
# table_data se inicializa como una lista vacía que almacenará los datos de la tabla en forma de lista de listas
table_data = []

# Iteramos sobre cada fila de la tabla
for row in rows:

  # Dentro de cada fila, se buscan todas las celdas (<td>: table data cells y <th>: table header), almacenándolas en cols
  cols = row.find_all(['td', 'th'])
  # Se utiliza una lista de comprensión para extraer el texto de cada celda, eliminando cualquier espacio en blanco al inicio y al final (strip=True)
  cols = [col.get_text(strip=True) for col in cols]
  # La lista de textos de celdas (cols) se agrega a table_data, construyendo así una lista de listas donde cada sublista representa una fila de la tabla
  table_data.append(cols)

table_data
```

```python
columnas = ['Jun 2025',
  'Jun 2024',
  'Change', '',
  'Programming Language',
  'Ratings',
  'Change']
data = table_data[1:]

# Crear el DataFrame
df = pd.DataFrame(data, columns=columnas)
df.head()
```

```python
df=df.drop(['Jun 2025',
  'Jun 2024',
  'Change', '',
  'Change'], axis=1)
df.head()
```

```python
df=df.replace({'%': ''}, regex=True)
df.head()
```
